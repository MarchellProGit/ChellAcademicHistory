```python
import numpy as np
import matplotlib.pyplot as plt
```

**1. Generate Data**

```python
jumlah_data = 100

y = np.array([i*0.1+np.random.randn() for i in range(jumlah_data)])
x = np.array([i*0.1 for i in range(jumlah_data)])
```

**10 Plot Data**

**2. Fungsi Linear**

```python
def ucup_linear(x,gradien) :
  y = gradien*x
  return y
```

Menghitung persamaan linear awal

```python
x_prediction = np.array([0,10])
m_prediction = 1
y_prediction = ucup_linear(x_prediction,m_prediction)
```

```python
plt.scatter(x, y)
plt.plot(x_prediction,y_prediction, "Red")
plt.axis([0,10,0,10])
plt.show()
```

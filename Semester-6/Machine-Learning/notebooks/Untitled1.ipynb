```python
import numpy as np
# Parameter Harmony Search
HMS = 10 # ukuran harmony memory
HMCR = 0.9 # harmony memory considering rate
PAR = 0.3 # pitch adjustment rate
NI = 100 # jumlah iterasi
area_size = 100
num_router = 3
num_node = 20
# Inisialisasi posisi node jaringan
nodes = np.random.rand(num_node, 2) * area_size
# Fungsi fitness (delay + cost + coverage)
def fitness(routers):
 delay = np.mean([np.min(np.linalg.norm(nodes - r, axis=1)) for r in routers])
 coverage = np.mean([np.any(np.linalg.norm(nodes - r, axis=1) < 20) for r in routers])
 return 0.7*delay + 0.3*(1 - coverage)
# Inisialisasi Harmony Memory
HM = [np.random.rand(num_router, 2) * area_size for _ in range(HMS)]
HM_fitness = [fitness(h) for h in HM]
for it in range(NI):
 new_harmony = []
 for r in range(num_router):
 if np.random.rand() < HMCR:
 idx = np.random.randint(HMS)
 pos = HM[idx][r]
 if np.random.rand() < PAR:
 pos += np.random.uniform(-1, 1, 2)
 else:
 pos = np.random.rand(2) * area_size
 new_harmony.append(pos)
 new_harmony = np.array(new_harmony)
 new_fit = fitness(new_harmony)
 # update harmony memory
 worst_idx = np.argmax(HM_fitness)
 if new_fit < HM_fitness[worst_idx]:
 HM[worst_idx] = new_harmony
 HM_fitness[worst_idx] = new_fit
best = HM[np.argmin(HM_fitness)]
print("Best router positions:\n", best)
print("Best fitness:", np.min(HM_fitness))
```

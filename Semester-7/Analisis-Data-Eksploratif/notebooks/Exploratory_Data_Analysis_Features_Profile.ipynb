```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import io

print("Menjalankan Blok Persiapan...")

# Nama file Excel Anda
file_name = "Canada.xlsx"
# Nama sheet di dalam file Excel
sheet_name_to_read = "Canada by Citizenship (2)"

try:
    # Menggunakan pd.read_excel()
    df = pd.read_excel(file_name, sheet_name=sheet_name_to_read)

    # --- PERSIAPAN UNTUK PLOTTING ---
    # Set 'OdName' (Nama Negara) sebagai index
    df_indexed = df.set_index('OdName')

    # == INI BAGIAN PENTING ==
    # Membuat DAFTAR TAHUN SEBAGAI ANGKA (INTEGER)
    years_2000_2013 = list(range(2000, 2014)) # Hasil: [2000, 2001, ..., 2013]

    print(f"Data '{file_name}' (Sheet: '{sheet_name_to_read}') berhasil dimuat.")
    print("Variabel 'df', 'df_indexed', dan 'years_2000_2013' (sebagai angka) siap digunakan.")

except FileNotFoundError:
    print(f"Error: File '{file_name}' tidak ditemukan.")
except ValueError as ve:
    print(f"Error: Sheet '{sheet_name_to_read}' tidak ditemukan di '{file_name}'.")
except Exception as e:
    print(f"Terjadi error saat memuat data: {e}")
```

```python
print("Membuat plot untuk Q5...")

# Negara yang akan di-plot
countries_q5 = ['Viet Nam', 'China', 'Algeria', 'India', 'Zambia']

try:
    # Pastikan 'df_indexed' dan 'years_2000_2013' sudah ada dari Blok Persiapan

    # Baris ini mengambil 'years_2000_2013' (sebagai angka) dari memori
    df_q5_data = df_indexed.loc[countries_q5, years_2000_2013].transpose()

    # --- Proses Plotting ---
    ax5 = df_q5_data.plot(kind='line', marker='o', figsize=(14, 8))

    ax5.set_title('Perbandingan dan Tren Jumlah Imigran (2000-2013)', fontsize=16)
    ax5.set_xlabel('Tahun', fontsize=12)
    ax5.set_ylabel('Jumlah Imigran', fontsize=12)
    ax5.legend(title='Negara', fontsize=10)
    ax5.grid(True)

    # Menyimpan gambar
    plt.tight_layout()
    fig5 = ax5.get_figure()
    fig5.savefig('q5_tren_imigran.png')
    plt.close(fig5)
    print("Plot Q5 berhasil disimpan sebagai 'q5_tren_imigran.png'")
    print("--- Sukses! ---")

except NameError:
    print("Error: Variabel 'df_indexed' atau 'years_2000_2013' tidak ditemukan.")
    print("Pastikan Anda sudah menjalankan 'Blok Persiapan' terlebih dahulu.")
except KeyError as e:
    # Error ini sekarang hanya akan muncul jika NAMA NEGARA salah ketik
    print(f"Error Q5: Salah satu negara tidak ditemukan. Cek ejaan: {e}")
except Exception as e:
    print(f"Error Q5: Terjadi error: {e}")
```

```python
import sys
import subprocess

try:
    import folium
    print("Pustaka 'folium' sudah terinstal.")
except ImportError:
    print("Pustaka 'folium' tidak ditemukan. Melakukan instalasi...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "folium"])
    import folium
    print("Pustaka 'folium' berhasil diinstal.")

print("Memastikan pustaka 'requests' terinstal dan mengunduh file GeoJSON...")
import requests

# URL file GeoJSON
geoj_url = "https://raw.githubusercontent.com/python-visualization/folium-example-data/main/world_countries.json"

try:
    # Mengunduh konten dari URL
    response = requests.get(geoj_url)
    response.raise_for_status() # Akan memunculkan HTTPError untuk kode status error

    # Memuat konten JSON
    world_geo = response.json()
    print("File GeoJSON berhasil diunduh dan dimuat ke variabel 'world_geo'.")

except requests.exceptions.RequestException as e:
    print(f"Error saat mengunduh file GeoJSON: {e}")
except Exception as e:
    print(f"Terjadi error tak terduga: {e}")
```

```python
print("Menghitung total imigran per negara dari 2000 hingga 2013...")

# 1. Pilih kolom tahun dari 2000 hingga 2013
df_years_selection = df_indexed[years_2000_2013]

# 2. Hitung jumlah total imigran untuk setiap negara selama periode tahun tersebut
total_immigrants_per_country = df_years_selection.sum(axis=1)

# 3. Konversi Series 'total_immigrants_per_country' menjadi DataFrame dengan kolom tunggal 'Total'
df_total_immigrants_2000_2013 = pd.DataFrame(total_immigrants_per_country, columns=['Total'])

print("Total imigran per negara dari 2000 hingga 2013 berhasil dihitung.")
print("DataFrame 'df_total_immigrants_2000_2013' dibuat dengan 5 baris pertama:")
print(df_total_immigrants_2000_2013.head())
```

```python
print("Ekstrak nama negara dari GeoJSON dan DataFrame imigrasi...")

# 1. Ekstrak nama negara dari world_geo (GeoJSON)
geojson_countries = [feature['properties']['name'] for feature in world_geo['features']]
print(f"Jumlah negara di GeoJSON: {len(geojson_countries)}")
# print("Contoh nama negara dari GeoJSON:", geojson_countries[:5]) # Un-comment to see examples

# 2. Ekstrak nama negara dari indeks df_total_immigrants_2000_2013
immigration_countries = df_total_immigrants_2000_2013.index.tolist()
print(f"Jumlah negara di DataFrame Imigrasi: {len(immigration_countries)}")
# print("Contoh nama negara dari DataFrame Imigrasi:", immigration_countries[:5]) # Un-comment to see examples

print("Kedua daftar nama negara berhasil diekstrak.")
```

```python
print("Membandingkan nama negara antara GeoJSON dan DataFrame imigrasi...")

# Temukan negara-negara di DataFrame imigrasi yang TIDAK ADA di GeoJSON
missing_in_geojson = [country for country in immigration_countries if country not in geojson_countries]
print(f"\nNegara di DataFrame Imigrasi tetapi TIDAK di GeoJSON ({len(missing_in_geojson)}):\n{missing_in_geojson}")

# Temukan negara-negara di GeoJSON yang TIDAK ADA di DataFrame imigrasi
missing_in_immigration = [country for country in geojson_countries if country not in immigration_countries]
print(f"\nNegara di GeoJSON tetapi TIDAK di DataFrame Imigrasi ({len(missing_in_immigration)}):\n{missing_in_immigration}")

# Beberapa nama umum untuk perbandingan langsung
print("\nBeberapa nama umum (jika ada):")
common_examples = ['United States', 'Republic of Korea', 'United Kingdom', 'Democratic Republic of the Congo']
for example in common_examples:
    if example in geojson_countries and example in immigration_countries:
        print(f"- {example} (ada di kedua sisi)")
    elif example in geojson_countries:
        print(f"- {example} (hanya di GeoJSON)")
    elif example in immigration_countries:
        print(f"- {example} (hanya di Imigrasi)")

print("Perbandingan nama negara selesai. Selanjutnya akan dibuat kamus pemetaan.")
```

```python
print("Menerapkan pemetaan nama negara ke DataFrame imigrasi...")

# Terapkan pemetaan ke indeks DataFrame
df_total_immigrants_2000_2013.rename(index=country_mapping, inplace=True)

print("Pemetaan nama negara berhasil diterapkan.")
print("5 baris pertama dari DataFrame setelah pemetaan nama negara:")
print(df_total_immigrants_2000_2013.head())
```

```python
print("Resetting index of df_total_immigrants_2000_2013 and renaming index column...")

# Reset index dan ubah nama kolom index menjadi 'Country'
df_total_immigrants_2000_2013_reset = df_total_immigrants_2000_2013.reset_index()
df_total_immigrants_2000_2013_reset.rename(columns={'OdName': 'Country'}, inplace=True)

print("Indeks berhasil direset dan kolom nama negara dinamai ulang menjadi 'Country'.")
print("5 baris pertama dari DataFrame setelah reset index:")
print(df_total_immigrants_2000_2013_reset.head())
```

```python
print("Membuat peta choropleth Folium...")

# 1. Buat objek peta Folium
# Titik tengah (latitude, longitude) dapat disesuaikan untuk memusatkan peta lebih baik
# Untuk peta dunia, (0,0) adalah pilihan umum, atau bisa mencoba (56, -97) untuk lebih dekat ke Kanada
canada_map = folium.Map(location=[0, 0], zoom_start=2, tiles='cartodbpositron')

# 2. Tambahkan lapisan choropleth ke peta
folium.Choropleth(
    geo_data=world_geo,
    data=df_total_immigrants_2000_2013_reset,
    columns=['Country', 'Total'],
    key_on='feature.properties.name',
    fill_color='YlGn',
    fill_opacity=0.7,
    line_opacity=0.2,
    legend_name='Total Immigrants to Canada (2000-2013)',
    highlight=True # Menyoroti negara saat kursor diarahkan
).add_to(canada_map)

# 3. Tambahkan lapisan kontrol untuk memudahkan interaksi
folium.LayerControl().add_to(canada_map)

# 4. Simpan peta sebagai file HTML
output_map_file = 'canada_immigration_choropleth_2000_2013.html'
canada_map.save(output_map_file)

print(f"Peta choropleth berhasil dibuat dan disimpan sebagai '{output_map_file}'.")
```

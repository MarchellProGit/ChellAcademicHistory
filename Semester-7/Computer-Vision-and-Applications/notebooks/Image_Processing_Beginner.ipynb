```python
import cv2
import matplotlib.pyplot as plt

# Load an image in color (RGB)
image_path = 'your_image.jpg'  # Replace with your image path
image_bgr = cv2.imread(image_path)
image_rgb = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

# Display the original RGB image
plt.figure(figsize=(6, 6))
plt.imshow(image_rgb)
plt.title("Original RGB Image")
plt.axis('off')
plt.show()

# Convert to grayscale
gray_image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2GRAY)

# Display grayscale image
plt.figure(figsize=(6, 6))
plt.imshow(gray_image, cmap='gray')
plt.title("Grayscale Image")
plt.axis('off')
plt.show()

# Split RGB channels
R, G, B = cv2.split(image_rgb)

# Display individual channels
fig, axs = plt.subplots(1, 3, figsize=(15, 5))
axs[0].imshow(R, cmap='Reds')
axs[0].set_title("Red Channel")
axs[0].axis('off')

axs[1].imshow(G, cmap='Greens')
axs[1].set_title("Green Channel")
axs[1].axis('off')

axs[2].imshow(B, cmap='Blues')
axs[2].set_title("Blue Channel")
axs[2].axis('off')

plt.show()

# Save processed images
cv2.imwrite('gray_image.jpg', gray_image)
cv2.imwrite('red_channel.jpg', R)
cv2.imwrite('green_channel.jpg', G)
cv2.imwrite('blue_channel.jpg', B)

print("Images saved successfully.")
```

```python
#Manual Gray Scale
import numpy as np
img = Image.open("/content/ronaldo-luis-nazario-de-lima-1472042256-5977.jpg").convert("RGB")
img_array = np.array(img)
gray_array = np.dot(img_array[..., :3], [0.299, 0.587, 0.114])
gray_img = Image.fromarray(gray_array.astype(np.uint8))
gray_img.save("manual_grayscale.jpg")
```

```python
#RGB Processing
from PIL import Image
import numpy as np
img = Image.open("/content/manual_grayscale.jpg").convert("RGB")
img_array = np.array(img)
r_channel = img_array.copy()
r_channel[:, :, 1] = 0  # Zero G
r_channel[:, :, 2] = 0  # Zero B
Image.fromarray(r_channel).save("red_channel.jpg")
# Repeat for G and B channels
```
